## Ejercicio 1. Lectura de conversaciones de un chatbot

**Contexto:** Una empresa ha desarrollado un chatbot basado en un LLM. Cada conversación se almacena en un CSV.

**Objetivo:** Mostrar todos los mensajes enviados por los usuarios.

**Datos de entrada:**
```
usuario,mensaje
u001,Hola
u002,¿Qué es Machine Learning?
u003,Necesito ayuda
```

**Resultado esperado:**
```
Hola
¿Qué es Machine Learning?
Necesito ayuda
```

**Pista:** Usa `csv.DictReader()`.

````python
import csv

# Datos de ejemplo
with open('conversaciones.csv', 'w', newline='', encoding='utf-8') as f:
    f.write('usuario,mensaje\n')
    f.write('u001,Hola\n')
    f.write('u002,¿Qué es Machine Learning?\n')
    f.write('u003,Necesito ayuda\n')

# Solución
with open('conversaciones.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        print(row['mensaje'])
````

## Ejercicio 2. Contabilizar mensajes procesados

**Contexto:** El responsable de IA necesita conocer el tamaño del conjunto de conversaciones.

**Objetivo:** Contar cuántos registros contiene el CSV.

**Resultado esperado:** `Número de mensajes: X`

**Pista:** Crea un contador e increméntalo dentro de un bucle.

````python
import csv

with open('conversaciones.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    contador = 0
    for row in reader:
        contador += 1

print(f'Número de mensajes: {contador}')
````

## Ejercicio 10. Analizar resultados de clasificación

**Contexto:** Un modelo de visión artificial clasifica imágenes.

**Objetivo:** Contar cuántas imágenes hay por categoría.

**Resultado esperado:**
```
gato: 3
perro: 2
```

**Reto opcional:** Mostrar también el porcentaje.

**Pista:** Usa `collections.Counter` o un diccionario.

````python
import csv
from collections import Counter

with open('imagenes.csv', 'w', newline='', encoding='utf-8') as f:
    f.write('imagen,categoria\n')
    f.write('img1.jpg,gato\n')
    f.write('img2.jpg,perro\n')
    f.write('img3.jpg,gato\n')
    f.write('img4.jpg,gato\n')
    f.write('img5.jpg,perro\n')

with open('imagenes.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    categorias = Counter(row['categoria'] for row in reader)

total = sum(categorias.values())
for cat, n in categorias.items():
    porcentaje = (n / total) * 100
    print(f'{cat}: {n} ({porcentaje:.1f}%)')
````

````python
import csv

total = spam = 0
with open('spam.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    for row in reader:
        total += 1
        if row['etiqueta'] == 'spam':
            spam += 1

no_spam = total - spam
porcentaje = (spam / total) * 100 if total else 0

print(f'Total mensajes: {total}')
print(f'Spam: {spam}')
print(f'No spam: {no_spam}')
print(f'Porcentaje spam: {porcentaje:.2f}%')
````

## Ejercicio 22. Conversión de JSON a CSV

**Objetivo:** Convertir un JSON en un CSV conservando todas las columnas.

````python
import csv, json

with open('inferencia.json', 'w', encoding='utf-8') as f:
    json.dump([
        {'texto': 'Hola', 'etiqueta': 'positivo', 'confianza': 0.92},
        {'texto': 'Adiós', 'etiqueta': 'negativo', 'confianza': 0.88}
    ], f, ensure_ascii=False)

with open('inferencia.json', encoding='utf-8') as f:
    datos = json.load(f)

columnas = list(datos[0].keys())
with open('inferencia.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=columnas)
    writer.writeheader()
    writer.writerows(datos)

print(open('inferencia.csv', encoding='utf-8').read())
````

## Ejercicio 24. Eliminación de registros duplicados

**Objetivo:** Crear `dataset_limpio.csv` sin duplicados.

````python
import csv

vistos = set()
filas_unicas = []
with open('spam_dups.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    columnas = reader.fieldnames
    for row in reader:
        clave = tuple(row.items())
        if clave not in vistos:
            vistos.add(clave)
            filas_unicas.append(row)

with open('dataset_limpio.csv', 'w', newline='', encoding='utf-8') as f:
    writer = csv.DictWriter(f, fieldnames=columnas)
    writer.writeheader()
    writer.writerows(filas_unicas)

print(f'{len(filas_unicas)} filas únicas guardadas')
print(open('dataset_limpio.csv', encoding='utf-8').read())
````

## Ejercicio 29. Pipeline automatizado de limpieza

**Objetivo:** Leer CSV, eliminar duplicados, detectar errores de estructura, crear copia limpia y generar informe.

````python
import csv

def pipeline(ruta_in, ruta_out, ruta_informe):
    filas_ok = []
    vistos = set()
    incidencias = {'duplicados': 0, 'corruptas': 0}
    with open(ruta_in, newline='', encoding='utf-8') as f:
        reader = csv.reader(f)
        cabecera = next(reader)
        for i, fila in enumerate(reader, 2):
            if len(fila) != len(cabecera):
                incidencias['corruptas'] += 1
                continue
            clave = tuple(fila)
            if clave in vistos:
                incidencias['duplicados'] += 1
                continue
            vistos.add(clave)
            filas_ok.append(fila)
    with open(ruta_out, 'w', newline='', encoding='utf-8') as f:
        w = csv.writer(f)
        w.writerow(cabecera)
        w.writerows(filas_ok)
    with open(ruta_informe, 'w', encoding='utf-8') as f:
        for k, v in incidencias.items():
            f.write(f'{k}: {v}\n')
    print('Incidencias:', incidencias)

pipeline('spam_dups.csv', 'limpio2.csv', 'informe.txt')
````

## Ejercicio 32. Preparación de datos para un sistema RAG

**Objetivo:** Leer documentos, detectar registros vacíos, eliminar duplicados, generar dataset limpio e informe. **Reto:** % descartados.

````python
import csv

with open('docs_rag.csv', 'w', newline='', encoding='utf-8') as f:
    f.write('id,texto\n')
    f.write('1,Documento uno\n')
    f.write('2,\n')
    f.write('3,Documento uno\n')
    f.write('4,Documento dos\n')

filas_ok, vistos = [], set()
vacios = dups = 0
with open('docs_rag.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    columnas = reader.fieldnames
    total = 0
    for row in reader:
        total += 1
        if not row['texto'].strip():
            vacios += 1
            continue
        k = row['texto']
        if k in vistos:
            dups += 1
            continue
        vistos.add(k)
        filas_ok.append(row)

with open('docs_rag_limpio.csv', 'w', newline='', encoding='utf-8') as f:
    w = csv.DictWriter(f, fieldnames=columnas)
    w.writeheader()
    w.writerows(filas_ok)

print(f'Total: {total}, vacíos: {vacios}, duplicados: {dups}')
print(f'% descartados: {(total-len(filas_ok))/total*100:.1f}%')
````

## Ejercicio 37. Monitorización de modelos en producción

**Objetivo:** Nº predicciones, distribución de etiquetas, variación respecto al día anterior, posibles anomalías.

````python
from collections import Counter

dia1 = ['positivo']*70 + ['negativo']*30
dia2 = ['positivo']*40 + ['negativo']*60

def distribucion(etiquetas):
    c = Counter(etiquetas)
    total = sum(c.values())
    return {k: v/total*100 for k, v in c.items()}, total

d1, n1 = distribucion(dia1)
d2, n2 = distribucion(dia2)

print(f'Día 1: {n1} preds -> {d1}')
print(f'Día 2: {n2} preds -> {d2}')

for etiqueta in d1:
    var = d2.get(etiqueta, 0) - d1[etiqueta]
    marca = '⚠ ANOMALÍA' if abs(var) > 20 else 'OK'
    print(f'Variación {etiqueta}: {var:+.1f}% {marca}')
````

## Ejercicio 40. Pipeline completo de datos para IA

**Objetivo:** Leer CSV, validar columnas, eliminar duplicados, detectar datos faltantes, generar estadísticas, crear dataset limpio y exportar informe.

**Entregables:** `dataset_limpio.csv`, `informe_validacion.csv`, `resumen_ejecutivo.txt`.

````python
import csv
from datetime import datetime

REQ = ['id', 'texto', 'etiqueta', 'confianza']

# Datos de entrada
with open('datos_brutos.csv', 'w', newline='', encoding='utf-8') as f:
    f.write('id,texto,etiqueta,confianza\n')
    f.write('1,Hola,positivo,0.9\n')
    f.write('2,Hola,positivo,0.9\n')  # duplicado
    f.write('3,,negativo,0.8\n')      # faltante
    f.write('4,Adiós,negativo,0.95\n')

with open('datos_brutos.csv', newline='', encoding='utf-8') as f:
    reader = csv.DictReader(f)
    columnas = reader.fieldnames
    if not all(c in columnas for c in REQ):
        raise ValueError('Faltan columnas obligatorias')
    filas_ok, vistos, inc = [], set(), {'duplicados': 0, 'faltantes': 0, 'total_leidas': 0}
    for row in reader:
        inc['total_leidas'] += 1
        clave = tuple(row.values())
        if clave in vistos:
            inc['duplicados'] += 1
            continue
        if any(not v.strip() for v in row.values()):
            inc['faltantes'] += 1
            continue
        vistos.add(clave)
        filas_ok.append(row)

# 1) dataset_limpio.csv
with open('dataset_limpio.csv', 'w', newline='', encoding='utf-8') as f:
    w = csv.DictWriter(f, fieldnames=REQ)
    w.writeheader()
    w.writerows(filas_ok)

# 2) informe_validacion.csv
with open('informe_validacion.csv', 'w', newline='', encoding='utf-8') as f:
    w = csv.writer(f)
    w.writerow(['metrica', 'valor'])
    for k, v in inc.items():
        w.writerow([k, v])
    w.writerow(['filas_validas', len(filas_ok)])

# 3) resumen_ejecutivo.txt
with open('resumen_ejecutivo.txt', 'w', encoding='utf-8') as f:
    f.write('RESUMEN EJECUTIVO DEL PIPELINE\n')
    f.write('==============================\n')
    f.write(f'Fecha: {datetime.now().isoformat()}\n')
    f.write(f'Filas leídas: {inc["total_leidas"]}\n')
    f.write(f'Duplicados eliminados: {inc["duplicados"]}\n')
    f.write(f'Registros con faltantes: {inc["faltantes"]}\n')
    f.write(f'Filas válidas: {len(filas_ok)}\n')

print('Pipeline completado.')
print(open('resumen_ejecutivo.txt', encoding='utf-8').read())
````
