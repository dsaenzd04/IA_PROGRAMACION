# ¿Qué es un CSV?

CSV significa **Comma Separated Values**.

Es un formato de almacenamiento de datos tabulares donde cada fila representa un registro y cada columna representa un atributo.

Es uno de los formatos más utilizados para intercambiar información entre herramientas de análisis de datos, bases de datos, hojas de cálculo y sistemas de inteligencia artificial.

## ¿Qué problema resuelve?

Permite almacenar información estructurada de forma sencilla.

Gracias a CSV podemos:

- Exportar datasets.
- Intercambiar datos entre aplicaciones.
- Compartir resultados de modelos.
- Almacenar métricas de entrenamiento.
- Registrar conversaciones de chatbots.
- Guardar experimentos de Machine Learning.

## ¿Cuándo se utiliza?

### Ciencia de Datos
Datasets de entrenamiento.

### NLP
Corpus de texto etiquetados.

### LLMs
Históricos de prompts y respuestas.

### Sistemas de recomendación
Registros de interacción.

### Visión artificial
Metadatos de imágenes.

### MLOps
Seguimiento de experimentos.

# Ventajas e inconvenientes

## Ventajas

- Muy sencillo.
- Amplio soporte.
- Fácil de generar.
- Fácil de leer.
- Compatible con Excel.
- Compatible con Python.
- Compatible con herramientas IA.

## Inconvenientes

- No permite estructuras anidadas.
- No conserva tipos de datos.
- Puede provocar problemas de codificación.
- Poco adecuado para información compleja.

# csv.reader()

## ¿Qué es?

Es una función que permite leer un fichero CSV fila a fila.

Cada registro se devuelve como una lista de Python.

## ¿Qué problema resuelve?

Permite procesar datasets sin tener que separar manualmente cada columna.

Esto es especialmente útil cuando trabajamos con grandes conjuntos de datos utilizados en Machine Learning.

## ¿Cuándo se utiliza?

- Lectura de datasets de entrenamiento.
- Lectura de logs de chatbots.
- Procesamiento de resultados de modelos.
- Carga de métricas de experimentación.
- Preparación de datos para análisis.

## Sintaxis general

```python
csv.reader(
    fichero,
    delimiter=',',
    quotechar='"'
)
```

# Ejemplo básico de csv.reader()

````python
import csv

with open('mensajes_spam.csv', encoding='utf-8') as fichero:
    lector = csv.reader(fichero)

    for fila in lector:
        print(fila)
````

### Salida esperada

```text
['id', 'mensaje', 'etiqueta']
['1', 'Gana dinero rapidamente', 'spam']
['2', 'Oferta exclusiva por tiempo limitado', 'spam']
['3', 'Reunion mañana a las 10', 'no_spam']
```

# Ignorar la cabecera

En la mayoría de proyectos profesionales no se procesa la fila de cabecera.

Para omitirla utilizamos `next()`.

````python
import csv

with open('mensajes_spam.csv', encoding='utf-8') as fichero:
    lector = csv.reader(fichero)

    next(lector)

    for fila in lector:
        print(fila)
````

# Ejemplo profesional

Supongamos que almacenamos conversaciones de un chatbot basado en un LLM.

````python
with open('chatbot_logs.csv', 'w', encoding='utf-8', newline='') as f:
    f.write('usuario,pregunta\n')
    f.write('u001,¿Qué es un LLM?\n')
    f.write('u002,Explícame Machine Learning\n')
````

````python
import csv

with open('chatbot_logs.csv', encoding='utf-8') as fichero:

    lector = csv.reader(fichero)

    next(lector)

    for fila in lector:
        usuario = fila[0]
        pregunta = fila[1]

        print(f'Usuario: {usuario}')
        print(f'Pregunta: {pregunta}')
        print('-' * 30)
````

# Parámetro delimiter

## ¿Qué es?

Define el carácter utilizado para separar columnas.

## ¿Cuándo se utiliza?

Muchos archivos exportados desde Excel utilizan punto y coma (`;`) en lugar de comas.

````python
with open('metricas.csv', 'w', encoding='utf-8') as f:
    f.write('modelo;accuracy\n')
    f.write('BERT;0.95\n')
    f.write('RoBERTa;0.96\n')
````

````python
import csv

with open('metricas.csv', encoding='utf-8') as fichero:

    lector = csv.reader(
        fichero,
        delimiter=';'
    )

    for fila in lector:
        print(fila)
````

# Parámetro quotechar

## ¿Qué es?

Indica el carácter utilizado para encerrar textos.

Es útil cuando los textos contienen comas.

````python
with open('comentarios.csv', 'w', encoding='utf-8') as f:
    f.write('texto,sentimiento\n')
    f.write('"Me encanta el producto, es fantástico",positivo\n')
````

````python
import csv

with open('comentarios.csv', encoding='utf-8') as fichero:

    lector = csv.reader(
        fichero,
        quotechar='"'
    )

    for fila in lector:
        print(fila)
````

# Parámetro newline

## ¿Qué es?

Controla los saltos de línea durante la escritura.

Cuando escribimos CSV siempre debemos utilizar:

```python
newline=''
```

Esto evita filas vacías inesperadas.

````python
with open(
    'salida.csv',
    'w',
    encoding='utf-8',
    newline=''
) as fichero:
    fichero.write('ok')
````

# csv.DictReader()

Hasta ahora hemos leído los datos utilizando índices de posición.

Sin embargo, en proyectos profesionales suele ser más cómodo acceder a los datos utilizando el nombre de la columna.

Para ello utilizamos `csv.DictReader()`.

## ¿Qué es?

Convierte cada fila del CSV en un diccionario de Python.

Las claves del diccionario corresponden a los nombres de las columnas.

## ¿Qué problema resuelve?

Evita tener que recordar posiciones numéricas.

Además hace el código más legible y menos propenso a errores cuando cambia la estructura del dataset.

## Sintaxis general

```python
csv.DictReader(fichero)
```

## Ejemplo básico

````python
import csv

with open('mensajes_spam.csv', encoding='utf-8') as fichero:
    lector = csv.DictReader(fichero)

    for fila in lector:
        print(fila)
````

### Salida esperada

```text
{'id': '1', 'mensaje': 'Gana dinero rapidamente', 'etiqueta': 'spam'}
{'id': '2', 'mensaje': 'Oferta exclusiva por tiempo limitado', 'etiqueta': 'spam'}
{'id': '3', 'mensaje': 'Reunion mañana a las 10', 'etiqueta': 'no_spam'}
```

## Acceso a valores concretos

````python
import csv

with open('mensajes_spam.csv', encoding='utf-8') as fichero:

    lector = csv.DictReader(fichero)

    for fila in lector:
        print(f'Mensaje: {fila["mensaje"]}')
        print(f'Etiqueta: {fila["etiqueta"]}')
        print()
````

# Ejemplo profesional

Supongamos un sistema de análisis de sentimientos utilizado para monitorizar la reputación online de una marca.

````python
with open('sentimientos.csv', 'w', encoding='utf-8', newline='') as fichero:
    fichero.write('texto,etiqueta,confianza\n')
    fichero.write('Excelente servicio,positivo,0.95\n')
    fichero.write('Muy mala experiencia,negativo,0.92\n')
````

````python
import csv

with open('metricas.csv', encoding='utf-8') as fichero:

    lector = csv.DictReader(fichero,delimiter=";")

    for fila in lector:

        accuracy = float(fila['accuracy'])

        if accuracy > 0.95:
            print(f"{fila['modelo']} supera el umbral de despliegue")
````

# Filtrado de registros

Una de las operaciones más habituales consiste en seleccionar únicamente los registros que cumplen determinadas condiciones.

````python
import csv

with open('sentimientos.csv', encoding='utf-8') as fichero:

    lector = csv.DictReader(fichero)

    for fila in lector:

        confianza = float(fila['confianza'])

        if confianza > 0.9:
            print(fila)
````

# Generación de un informe resumido

Los científicos de datos generan constantemente informes a partir de los datasets procesados.

````python
import csv

total_registros = 0
total_spam = 0

with open('mensajes_spam.csv', encoding='utf-8') as fichero:

    lector = csv.DictReader(fichero)

    for fila in lector:

        total_registros += 1

        if fila['etiqueta'] == 'spam':
            total_spam += 1

print('INFORME')
print('-------')
print('Total registros:', total_registros)
print('Total spam:', total_spam)
````

# Caso real de MLOps

Durante el entrenamiento de modelos es habitual almacenar información sobre cada ejecución.

Estos datos posteriormente pueden enviarse a sistemas como MLflow, Weights & Biases o plataformas internas de seguimiento.

````python
import csv

experimento = {
    'modelo': 'BERT',
    'accuracy': 0.96,
    'precision': 0.95,
    'recall': 0.94,
    'f1': 0.95
}

with open('tracking_experimentos.csv', 'w', encoding='utf-8', newline='') as fichero:

    writer = csv.DictWriter(
        fichero,
        fieldnames=experimento.keys()
    )

    writer.writeheader()
    writer.writerow(experimento)

print('Experimento registrado correctamente')
````

# Ejercicio 3. Mostrar únicamente los mensajes spam

## Contexto

Se está construyendo un detector de spam para correo electrónico.

## Objetivo

Mostrar únicamente los registros etiquetados como `spam`.

## Datos de entrada

```csv
mensaje,etiqueta
Oferta exclusiva,spam
Buenos días,no_spam
Gana dinero rápido,spam
```

## Resultado esperado

```text
Oferta exclusiva
Gana dinero rápido
```

## Pista

Utiliza una estructura `if`.

````python
import csv

with open("ejercicio3.csv", "w",encoding="utf-8",newline="") as fichero:
    fichero.write("message,tag\n")
    fichero.write("Oferta exclusiva,spam\n")
    fichero.write("Buenos días,no_spam\n")
    fichero.write("Gana dinero rápido,spam\n")
````
