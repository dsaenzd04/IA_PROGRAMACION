Luego, guárdalo en un archivo llamado 'dataset_ia.jsonl

```python
import json


def guardar_lista_mensajes(ruta,lista):
    """Guarda lista de mensajes en un archivo JSON."""
    messages = []

    for diccionario in lista:
      for clave, valor in diccionario.items():
        message = {
            "role" : clave,
            "content" : valor,
        }
        messages.append(message)

    with open(ruta, "w", encoding="utf-8") as f:
      json.dump({"messages": messages}, f,indent=4, ensure_ascii=False)

def leer_json():
  with open("dataset_ia.jsonl", encoding="utf-8") as f:
      print(json.load(f))
```

```python
conversaciones = [
    {"usuario": "¿Qué es un vector?",
     "asistente": "En IA, es una lista de números que representa el significado de un texto."},
    {"usuario": "Dame un ejemplo de embedding",
     "asistente": "El texto 'perro' podría ser [0.23, -0.45, 0.89]."}
]

guardar_lista_mensajes("dataset_ia.jsonl", conversaciones)

"El archuivo se guarda en archivos del Colab"
leer_json()
```
